### Message Prompt

Message are fundamental unit of context for AI model in langchain.They represent the input output of model,carrying both context and metadata.
Message are Object that contains:
  *Role:Identify the message type (e.g system,user)
  *Content:The actual text or data of the message 
  *Additional metadata:such as name,additional kwargs 

Message types

1.System message : Tell the model how to behave and provide the context for interaction
2.Human Message : Represent user input and interact with ai model
3.AI Message : Response generated by model,including text content,tool calls and metadata
4.Tool Message : Represent the output of tool calls

```python
import os
from langchain_groq import ChatGroq
model=ChatGroq(model="openai/gpt-oss-20b")
```

```python
from langchain.messages import SystemMessage,HumanMessage

messages=[
    SystemMessage("You are poetry expert"),
    HumanMessage("Write a poem about AI")
]

response=model.invoke(messages)
response
```



### Dict role ==>	LangChain class

## "user"     ==>  HumanMessage
## "assistant" ==>  AIMessage
## "tool"      ==>     ToolMessage

```python
messages = [
    {"role": "system", "content": "You are helpful."},
    {"role": "user", "content": "Hello!"},
    {"role": "assistant", "content": "Hi! How can I help?"},
    {"role": "user", "content": "What is LangChain?"},
]
response = model.invoke(messages)
response
```
