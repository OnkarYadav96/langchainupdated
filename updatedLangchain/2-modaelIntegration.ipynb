### Stream and Batch

### Stream

Streaming means the model sends the answer piece by piece as it is generated, instead of waiting until the full response is ready. This is useful for real-time applications where you want to show progress or start processing parts of the response as soon as they are available.
For that we use stream() method

```python
import os
from langchain_groq import ChatGroq


model=ChatGroq(model="openai/gpt-oss-20b")
for chunk in model.stream("Explain AI in 30 sentences"):
    print(chunk.content, end="", flush=True)  # print each piece immediately
```

### Batch
Instead of sending one prompt → wait → one answer, you send many prompts and get many answers in one organized run.

```python
questions = [
    "What is AI?",
    "What is ML?",
    "What is deep learning?",
]

responses = model.batch(questions)
for q, r in zip(questions, responses):
    print(q, "→", r.content)
```
